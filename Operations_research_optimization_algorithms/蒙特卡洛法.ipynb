## 蒙特卡洛法求圆周率：

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import rcParams

# 设置中文字体，比如 SimHei（黑体），或微软雅黑
rcParams['font.sans-serif'] = ['SimHei']   # 或者 ['Microsoft YaHei']
rcParams['axes.unicode_minus'] = False     # 避免负号显示成方块


# 参数初始化：投放10000个点，圆半径为1，圆心坐标（1，1）
p = 100000 # 总共要投放的点数
r = 1 # 圆的半径
x0, y0 = 1, 1 # 圆心坐标
n = 0 # 初始还未投放点，有0个点在圆内

# 绘制窗口
plt.figure()
plt.title('蒙特卡洛法求圆周率')
plt.xlabel('x')
plt.ylabel('y')
# 保持绘图窗口，多次绘图
for i in range(p):
    # np.random.rand()函数产生在（0，1）之间的随机数
    px = np.random.rand() * 2 # 随机生成点的横坐标
    py = np.random.rand() * 2 # 随机生成点的纵坐标

    # 判断点是否在圆内
    if (px - x0) ** 2 + (py - y0) ** 2 < r ** 2:
        plt.plot(px, py, '.', color='b') # 圆内点用蓝色表示
        n += 1 # n用于计算圆内点的总数
    else:
        plt.plot(px, py, '.', color='r') # 圆外的点用红色表示

plt.axis('equal') # 绘图时横纵坐标单位长度相同，便于观察图
plt.show()

# 计算圆周率的估计值
s = (n / p) * 4
pi = s / (r ** 2)
print("pi =", pi)
```
